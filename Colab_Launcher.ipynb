### 🚀 **MERGING STUDIO v2.0 — 1-CELL GITHUB & GRADIO LAUNCHER**
Sel tunggal ini akan meng-clone repositori **Merging Studio v2.0** dari GitHub Anda, memasang dependensi sistem (`aria2`, `safetensors`, `gradio`), menjalankan pengujian otomatis *Golden Layer Invariants*, dan membuka tautan **Gradio Web Studio (`*.gradio.live`)**.

```python
# @title **▶️ 1. Clone GitHub Repo & Launch Merging Studio v2.0 (Gradio Live Link)**
import os
import subprocess

# @markdown ### **🔗 Konfigurasi Repositori GitHub Anda**
github_repo_url = "https://github.com/SEVUNX/Merging-Studio-v2.git" # @param {type:"string"}
branch_name = "main" # @param {type:"string"}
mount_google_drive = False # @param {type:"boolean"}
run_invariant_self_test = True # @param {type:"boolean"}

os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"
target_dir = "/content/Merging-Studio-v2"

print("⚙️ [1/4] Checking system dependencies (aria2c, git-lfs, Python packages)...")
if not os.path.exists("/usr/bin/aria2c"):
    subprocess.run("apt-get -y install -qq aria2 git-lfs > /dev/null", shell=True)
subprocess.run("pip install -q safetensors torch numpy ml_dtypes tqdm huggingface_hub gradio fastapi uvicorn", shell=True)

if mount_google_drive:
    from google.colab import drive
    print("🚀 Mounting Google Drive at /content/drive...")
    drive.mount("/content/drive")

print(f"🐙 [2/4] Syncing repository from {github_repo_url} ({branch_name})...")
if not os.path.exists(target_dir):
    subprocess.run(["git", "clone", "-b", branch_name, github_repo_url, target_dir])
else:
    subprocess.run(["git", "-C", target_dir, "pull", "origin", branch_name])

if run_invariant_self_test and os.path.exists(f"{target_dir}/tests/test_invariants.py"):
    print("🛡️ [3/4] Running Golden Layer Invariant Self-Test...")
    subprocess.run(["python3", f"{target_dir}/tests/test_invariants.py"], check=True)

print("🌐 [4/4] Starting Merging Studio v2.0 Web UI (Click the *.gradio.live link below)...")
!python3 /content/Merging-Studio-v2/app.py --share
```
